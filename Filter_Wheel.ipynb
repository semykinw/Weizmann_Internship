```python
#Time test for a Filter Wheel
import serial
import time
import re

fw = serial.Serial("COM25", baudrate=115200, timeout=2)
time.sleep(2)

def get_position(fw):
    fw.reset_input_buffer()
    fw.write(b"pos?\r")
    time.sleep(0.1)
    raw = fw.read(fw.in_waiting)
    text = raw.decode(errors='ignore')
    # ищем число (может быть двузначным)
    match = re.search(r'\d+', text)
    if match:
        return match.group()
    return None

# Встаём на позицию 1
fw.reset_input_buffer()
fw.write(b"pos=1\r")
print("Переходим на позицию 1...")
for _ in range(50):
    pos = get_position(fw)
    print(f"  текущая позиция: {pos}")
    if pos == "1":
        break
    time.sleep(0.2)
print("Позиция 1 достигнута. Жду 1 секунду...")
time.sleep(1)

# Проходим позиции 2 → 12
t_start = time.perf_counter()

for target in range(2, 13):
    fw.reset_input_buffer()
    fw.write(f"pos={target}\r".encode())

    for _ in range(100):
        pos = get_position(fw)
        print(f"  опрос: {pos}")
        if pos == str(target):
            t_end = time.perf_counter()
            break
        time.sleep(0.1)
    else:
        print(f"Не удалось достичь позиции {target}")
        fw.close()
        exit()

    print(f"Переход {target-1}→{target}: {t_end - t_start:.3f} сек")
    t_start = time.perf_counter()

fw.close()
```
