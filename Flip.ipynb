```python
import time

MEASURE_DURATION = 1 * 60    # 1 минута в секундах
PAUSE_DURATION   = 20 * 60   # 20 минут в секундах
TOTAL_DURATION   = 10 * 60 * 60  # 10 часов в секундах

t_global_start = time.time()
cycle = 1

while (time.time() - t_global_start) < TOTAL_DURATION:

    # --- Измерение 1 минута ---
    print(f"Цикл {cycle}: начало измерения — {time.strftime('%H:%M:%S')}")
    sys.ctrl(flipperA='down')
    time.sleep(MEASURE_DURATION)
    sys.ctrl(flipperA='up')
    print(f"Цикл {cycle}: измерение завершено — {time.strftime('%H:%M:%S')}")

    # Проверяем не истекли ли 10 часов
    if (time.time() - t_global_start) >= TOTAL_DURATION:
        break

    # --- Пауза 20 минут ---
    print(f"Цикл {cycle}: пауза 20 минут — {time.strftime('%H:%M:%S')}")
    time.sleep(PAUSE_DURATION)

    cycle += 1

print(f"Эксперимент завершён — {time.strftime('%H:%M:%S')}")
```
